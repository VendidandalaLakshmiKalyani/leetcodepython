```python
def sortColors(nums):
    count0 = 0
    count1 = 0
    count2 = 0

    # Count 0s, 1s and 2s
    for i in range(len(nums)):
        if nums[i] == 0:
            count0 += 1
        elif nums[i] == 1:
            count1 += 1
        else:
            count2 += 1

    # Place all 0s
    i = 0
    while count0 > 0:
        nums[i] = 0
        i += 1
        count0 -= 1

    # Place all 1s
    while count1 > 0:
        nums[i] = 1
        i += 1
        count1 -= 1

    # Place all 2s
    while count2 > 0:
        nums[i] = 2
        i += 1
        count2 -= 1


# Example input
nums = [2, 0, 2, 1, 1, 0]

sortColors(nums)

print(nums)
```
