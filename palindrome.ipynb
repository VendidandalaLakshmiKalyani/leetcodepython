```python
x = 121
original = x
reversed_num = 0

print("Original number:", x)

if x < 0:
    print("Negative number cannot be a palindrome")
    print("Output: False")
else:
    while x > 0:
        digit = x % 10
        print("Last digit:", digit)

        reversed_num = reversed_num * 10 + digit
        print("Reversed number:", reversed_num)

        x = x // 10
        print("Remaining number:", x)

    if original == reversed_num:
        print("Output: True")
    else:
        print("Output: False")
```
